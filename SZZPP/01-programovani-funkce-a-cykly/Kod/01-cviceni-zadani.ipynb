## Úloha 2 — Odhad splátek půjčky

*Archetyp: nekonečný while (katalog #6)*

### Chybný kód

```python
# Funkce má spočítat, kolik měsíců potrvá splatit půjčku
# při dané měsíční splátce a úroku.

def pocet_splatek(dluh, splatka, urok_mesicne=0.01):
    mesice = 0
    while dluh > 0:
        dluh = dluh * (1 + urok_mesicne)
        dluh - splatka
    return mesice


# POZOR: tenhle řádek se zacyklí! Než ho odkomentuješ, měj po ruce Interrupt kernel.
# print(pocet_splatek(10000, 500))
```

### Rozšiř o následující funkčnost

- vrať dvojici `(počet měsíců, celkem zaplaceno)`
- ošetři případ, kdy je **splátka menší než měsíční úrok** — dluh nikdy nesplatíš,
  vyhoď `ValueError` místo zacyklení
- přidej pojistku `max_mesicu=600`, po jejímž překročení se cyklus ukončí
- ošetři nekladný dluh i splátku

## Úloha 3 — Zařazení do věkové kategorie

*Archetyp: pořadí elif, nedosažitelná větev (katalog #7)*

### Chybný kód

```python
# Funkce má vrátit sportovní kategorii podle věku.

def kategorie(vek):
    if vek >= 6:
        return "žák"
    elif vek >= 15:
        return "dorostenec"
    elif vek >= 19:
        return "dospělý"
    elif vek >= 60:
        return "veterán"
    else:
        return "předškolák"


for v in [4, 8, 16, 25, 65]:
    print(f"{v:3} let -> {kategorie(v)}")
```

### Rozšiř o následující funkčnost

- ošetři **záporný věk** a nečíselný vstup (`ValueError` / `TypeError`)
- přidej pojmenovaný parametr `hranice` — slovník kategorie→minimální věk,
  aby šly hranice změnit zvenčí
- přidej funkci `kategorie_rozsah(nazev)` vracející dvojici `(od, do)` pro danou kategorii

## Úloha 4 — Výpis násobilky

*Archetyp: off-by-one v range (katalog #8)*

### Chybný kód

```python
# Funkce má vypsat násobilku daného čísla od 1 do n
# a vrátit součet všech vypsaných násobků.

def nasobilka(cislo, n=10):
    soucet = 0
    for i in range(1, n):
        nasobek = cislo * i
        print(f"{cislo} x {i} = {nasobek}")
        soucet += nasobek
    return soucet


print("součet:", nasobilka(3))
# čekáme 3+6+9+...+30 = 165
```

### Rozšiř o následující funkčnost

- přidej pojmenovaný parametr `krok=1`
- přidej parametr `od=1`, aby šlo začít jinde než od jedničky
- vrať **dvojici** `(součet, počet vypsaných řádků)`
- ošetři nekladné `n` a nulový nebo záporný krok

## Úloha 5 — Zápis do docházky

*Archetyp: mutable default argument (katalog #10)*

### Chybný kód

```python
# Funkce zapíše jméno do seznamu přítomných.
# Když seznam nedostane, založí nový.

def zapis(jmeno, dochazka=[]):
    dochazka.append(jmeno)
    return dochazka


pondeli = zapis("Anna")
print("pondělí:", pondeli)

utery = zapis("Bob")
print("úterý:  ", utery)      # čekáme ['Bob'], ale…
```

### Rozšiř o následující funkčnost

- ať funkce **nezapíše duplicitu** (stejné jméno dvakrát)
- přidej funkci `odepis(dochazka, jmeno)` — neznámé jméno vyhodí `ValueError`
- přidej `souhrn(dochazka)` vracející formátovaný řetězec s počtem a jmény
- ošetři prázdné jméno a nesprávný typ

## Úloha 6 — Hledání odmocniny půlením intervalu

*Archetyp: porovnání float, while bez tolerance (katalog #11, #6)*

### Chybný kód

```python
# Funkce má najít odmocninu čísla metodou půlení intervalu.

def odmocnina(cislo):
    dolni = 0
    horni = cislo
    stred = (dolni + horni) / 2

    while stred * stred != cislo:
        if stred * stred < cislo:
            dolni = stred
        else:
            horni = stred
        stred = (dolni + horni) / 2

    return stred


# POZOR: pro většinu vstupů se zacyklí!
# print(odmocnina(16))   # 4 se najde
# print(odmocnina(2))    # zacyklí se — √2 je iracionální
```

### Rozšiř o následující funkčnost

- přidej pojmenovaný parametr `tolerance=1e-9`
- přidej pojistku na maximální počet iterací
- ošetři **záporné číslo** (`ValueError`) a čísla **menší než 1**
  (pro `cislo < 1` je odmocnina větší než `cislo`, takže horní mez nestačí!)
- vrať dvojici `(odmocnina, počet iterací)` a porovnej s `math.sqrt`
