## Úloha 1 — Odstranění záporných hodnot

*Archetyp: mazání během iterace (katalog #6)*

### Chybný kód

```python
# Funkce má ze seznamu měření odstranit všechny záporné hodnoty
# (chybná měření ze senzoru) a vrátit počet odstraněných.

def uklid_mereni(mereni):
    odstraneno = 0
    print(mereni)
    kopie = mereni
    print(kopie)
    for hodnota in mereni:
        print(hodnota)
        if hodnota < 0:
            print(kopie)
            odstraneno += 1
    return odstraneno


data = [12, -1, -3, 15, 20, -7, 18]
print(uklid_mereni(data))
print(data)
```

### Rozšiř o následující funkčnost

- místo modifikace vstupu vrať **novou dvojici** `(očištěný seznam, počet odstraněných)`
- funkce ať přijímá volitelný **pojmenovaný parametr `mez=0`** — odstraní se vše pod mezí
- ošetři, že vstup je seznam čísel, jinak výjimka

## Úloha 2 — Počítání četností slov

*Archetyp: KeyError u slovníku (katalog #12, #14)*

### Chybný kód

```python
# Funkce má spočítat, kolikrát se ve větě vyskytuje které slovo.

def cetnosti(veta):
    pocty = {}
    for slovo in veta.split():
        pocty[slovo] += 1
    return pocty


text = "pes kocka pes myš kocka pes"
vysledek = cetnosti(text)

for slovo, pocet in vysledek:
    print(f"{slovo}: {pocet}")
```

### Rozšiř o následující funkčnost

- ignoruj velikost písmen (`Pes` a `pes` je totéž)
- přidej pojmenovaný parametr `min_vyskyt=1` — vrátí jen slova s aspoň tolika výskyty
- vrať slova **seřazená sestupně podle četnosti**
- prázdný vstup ať vyhodí `ValueError`

## Úloha 3 — Kopie rozvrhu

*Archetyp: mělká vs. hluboká kopie (katalog #7, #8)*

### Chybný kód

```python
# Funkce má vytvořit zálohu rozvrhu, aby šlo v původním bezpečně měnit.
# Rozvrh je slovník: den -> seznam předmětů.

import copy

def zaloha(rozvrh):
    kopie = rozvrh
    return kopie


rozvrh = {"po": ["APR", "MAT"], "ut": ["OPS"]}
zal = zaloha(rozvrh)

rozvrh["po"].append("ZEL")
rozvrh["st"] = ["URDB"]

print("original:", rozvrh)
print("zaloha:  ", zal)   # měla by být beze změny!
```

### Rozšiř o následující funkčnost

- ukaž **tři varianty** vedle sebe: `= rozvrh`, `.copy()`, `deepcopy` — a co která zvládne
- napiš funkci `porovnej(a, b)`, která vrátí seznam dnů, kde se rozvrhy liší
- ošetři, že vstup je slovník se seznamy hodnot, jinak výjimka

## Úloha 4 — Průměr každého druhého měření

*Archetyp: index vs. hodnota, off-by-one (katalog #4, #11)*

### Chybný kód

```python
# Funkce má spočítat průměr měření na SUDÝCH pozicích
# (první, třetí, páté … tedy indexy 0, 2, 4).

def prumer_sudych_pozic(mereni):
    soucet = 0
    pocet = 0
    for m in mereni:
        if m % 2 == 0:
            soucet += m
            pocet += 1
    return soucet / pocet


data = [10, 3, 20, 7, 30]
print(prumer_sudych_pozic(data))   # čekáme (10+20+30)/3 = 20.0
```

### Rozšiř o následující funkčnost

- přidej parametr `krok=2`, aby šlo brát každý n-tý prvek
- vrať dvojici `(průměr, počet započtených)`
- ošetři **prázdný seznam** i seznam, kde po výběru nezůstane nic (dělení nulou)
- napiš druhou variantu **přes slicing** a ověř, že dává totéž

## Úloha 5 — Sbírání položek do košíku

*Archetyp: mutable default argument (katalog #13)*

### Chybný kód

```python
# Funkce přidá položku do košíku. Když košík nedostane, založí nový.

def pridej(polozka, kosik=[]):
    kosik.append(polozka)
    return kosik


kosik_anny = pridej("chleba")
print("Anna:", kosik_anny)

kosik_boba = pridej("mléko")
print("Bob: ", kosik_boba)     # čekáme ['mléko'], ale…
```

### Rozšiř o následující funkčnost

- přidej parametr `mnozstvi=1`, košík ať je **slovník** položka → počet
- funkce `odeber(kosik, polozka)`, která sníží počet a při nule klíč smaže;
  neexistující položka ať vyhodí `KeyError` s vlastní hláškou

## Úloha 6 — Seřazení a normalizace jmen

*Archetyp: metody vracející None (katalog #9, #10)*

### Chybný kód

```python
# Funkce má vzít seznam jmen, oříznout mezery, sjednotit velikost písmen
# (první velké), seřadit abecedně a vrátit výsledek.

def uprav_jmena(jmena):
    ocistena = []
    for j in jmena:
        ocistena.append(j.strip().capitalize())
    ocistena = ocistena.sort()
    return ocistena


vstup = ["  novák ", "DVOŘÁK", "  Černý"]
print(uprav_jmena(vstup))
```

### Rozšiř o následující funkčnost

- přidej parametr `sestupne=False` pro obrácené řazení
- odstraň **duplicity** (po normalizaci) a zachovej abecední pořadí
- přidej parametr `min_delka=1`, který kratší jména vyhodí
- prázdný řetězec nebo jiný typ než `str` ať vyhodí `ValueError`
