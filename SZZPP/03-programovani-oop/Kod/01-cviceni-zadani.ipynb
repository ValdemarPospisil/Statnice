## Úloha 1 — Bankovní účet

*Archetyp: chybějící self, odsazení metod (katalog #1, #2)*

### Chybný kód

```python
# Třída má reprezentovat bankovní účet s vkladem a výběrem.

class Ucet:
    def __init__(majitel, zustatek=0):
        self.majitel = majitel
        self.zustatek = zustatek

        def vloz(castka):
            self.zustatek += castka

        def vyber(castka):
            self.zustatek -= castka


u = Ucet("Anna", 1000)
u.vloz(500)
print(u.zustatek)
```

### Rozšiř o následující funkčnost

- ošetři **záporný vklad i výběr** (`ValueError`)
- výběr přes zůstatek ať vyhodí `ValueError` s informací, kolik chybí
- přidej `__str__` a `__repr__`
- přidej **historii transakcí** (seznam) a metodu `vypis_historii()`

## Úloha 2 — Počítadlo návštěv

*Archetyp: atribut třídy vs. instance (katalog #13, #3)*

### Chybný kód

```python
# Třída má počítat návštěvy jednotlivých stránek webu.

class Stranka:
    navstevy = []          # sem se ukládají časy návštěv

    def __init__(self, url):
        self.url = url

    def navstiv(self, cas):
        self.navstevy.append(cas)

    def pocet(self):
        return len(navstevy)


home = Stranka("/")
about = Stranka("/about")

home.navstiv("10:00")
home.navstiv("10:05")
about.navstiv("11:00")

print("home:", home.pocet())      # čekáme 2
print("about:", about.pocet())    # čekáme 1
```

### Rozšiř o následující funkčnost

- přidej **atribut třídy `celkem_stranek`**, který počítá, kolik instancí vzniklo
  (tohle je legitimní použití atributu třídy!)
- přidej `__len__`, aby `len(stranka)` vrátilo počet návštěv
- přidej `__contains__`, aby `"10:00" in stranka` fungovalo
- ošetři neplatné URL (musí začínat `/`)

## Úloha 3 — Zlomek

*Archetyp: __str__ vs. __repr__ (katalog #7, #18)*

### Chybný kód

```python
# Třída má reprezentovat zlomek a umět ho hezky vypsat.

class Zlomek:
    def __init__(self, citatel, jmenovatel):
        self.citatel = citatel
        self.jmenovatel = jmenovatel

    def __str__(self):
        f"{self.citatel}/{self.jmenovatel}"


z = Zlomek(3, 4)
print(z)
print([z, z])
print(f"zlomek je {z}")
```

### Rozšiř o následující funkčnost

- ošetři **nulový jmenovatel** (`ZeroDivisionError` nebo `ValueError`)
- přidej `__repr__`
- **zkrať zlomek** v konstruktoru na základní tvar (`math.gcd`)
- zápornost ať se drží v čitateli (`1/-2` → `-1/2`)
- přidej `hodnota` jako **property** vracející desetinné číslo

## Úloha 4 — Barva v paletě

*Archetyp: __eq__ bez __hash__ (katalog #6, #15)*

### Chybný kód

```python
# Třída má reprezentovat RGB barvu a umět porovnávat barvy.

class Barva:
    def __init__(self, r, g, b):
        self.r, self.g, self.b = r, g, b

    def __eq__(self, other):
        return self.r == other.r and self.g == other.g and self.b == other.b

    def __repr__(self):
        return f"Barva({self.r}, {self.g}, {self.b})"


cervena = Barva(255, 0, 0)
zelena = Barva(0, 255, 0)

print(cervena == Barva(255, 0, 0))     # True

paleta = {cervena, zelena}             # unikátní barvy
print(paleta)
```

### Rozšiř o následující funkčnost

- ošetři složky **mimo rozsah 0–255** i nesprávný typ
- přidej `__hash__`, aby šla barva do množiny a jako klíč slovníku
- přidej property `jas` (0–1) podle vzorce $0.299r + 0.587g + 0.114b$ děleno 255
- přidej property `je_tmava` (jas pod 0.5)
- přidej `hex` jako property vracející `#ff0000`

## Úloha 5 — Teploměr

*Archetyp: rekurze v property, chybějící setter (katalog #9, #8)*

### Chybný kód

```python
# Třída má hlídat, aby teplota nešla pod absolutní nulu.

class Teplomer:
    def __init__(self, celsia):
        self.celsia = celsia

    @property
    def celsia(self):
        return self.celsia

    @celsia.setter
    def celsia(self, hodnota):
        if hodnota < -273.15:
            raise ValueError("pod absolutní nulou")
        self.celsia = hodnota

    @property
    def fahrenheity(self):
        return self.celsia * 9 / 5 + 32


t = Teplomer(20)
print(t.celsia)
```

### Rozšiř o následující funkčnost

- přidej property `kelviny` (jen ke čtení)
- umožni **zapisovat i ve fahrenheitech** (setter, který přepočítá na Celsia)
- přidej property `stav` vracející `"mráz"` / `"chladno"` / `"teplo"` / `"horko"`
- přidej `__str__` s formátováním na jedno desetinné místo

## Úloha 6 — Kruhový buffer

*Archetyp: iterátor bez StopIteration (katalog #10, #11, #12)*

### Chybný kód

```python
# Třída má procházet posledních N záznamů dokola.

class Buffer:
    def __init__(self, polozky):
        self.polozky = polozky

    def __iter__(self):
        self.i = 0
        return self.polozky

    def __next__(self):
        hodnota = self.polozky[self.i]
        self.i += 1
        return hodnota


b = Buffer(["a", "b", "c"])
for x in b:
    print(x)
```

### Rozšiř o následující funkčnost

- oprav iterátor tak, aby prošel položky **právě jednou**
- přidej `__len__` a `__contains__`
- přidej metodu `pridej(polozka)` s **omezenou kapacitou** — nejstarší vypadne
- přidej **druhou variantu iterátoru jako generátor** a porovnej chování
- ošetři prázdný buffer
